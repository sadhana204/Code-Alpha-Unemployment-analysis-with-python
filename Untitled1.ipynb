# Unemployment Analysis with Python-

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option('display.max_columns', None)
sns.set_theme(style='whitegrid')
print('Libraries imported successfully!')
```

# Load the Dataset-

```python
df = pd.read_csv('/content/Unemployment_Rate_upto_11_2020.csv')
```

# Initial Data Inspection-

```python
print('Shape:', df.shape)
print('\nFirst 5 rows:')
display(df.head())

print('\nColumn names:')
print(df.columns.tolist())

print('\nData types:')
print(df.dtypes)
```

# Data Cleaning-

```python
df.columns = [str(c).strip() for c in df.columns]

# Rename the second "Region" column to Zone.
if len(df.columns) >= 7 and df.columns[6] == 'Region':
    df.columns.values[6] = 'Zone'

df['Date'] = pd.to_datetime(df['Date'], dayfirst=True, errors='coerce')

numeric_cols = [
    'Estimated Unemployment Rate (%)',
    'Estimated Employed',
    'Estimated Labour Participation Rate (%)'
]

for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')

print('Missing values before cleaning:')
display(df.isnull().sum())

df = df.dropna(subset=['Region', 'Date'] + numeric_cols).copy()
df = df.sort_values('Date').reset_index(drop=True)

print('\nShape after cleaning:', df.shape)
display(df.head())
```

# Descriptive Statistics-

```python
display(df[numeric_cols].describe().round(2))

print('Number of states/regions:', df['Region'].nunique())
print('Date range:', df['Date'].min().date(), 'to', df['Date'].max().date())
```

# Monthly Unemployment Trend-

```python
monthly_rate = df.groupby('Date')['Estimated Unemployment Rate (%)'].mean().reset_index()

plt.figure(figsize=(10,5))
sns.lineplot(data=monthly_rate, x='Date', y='Estimated Unemployment Rate (%)', marker='o')
plt.title('Average Unemployment Rate in India — Monthly Trend')
plt.xlabel('Month')
plt.ylabel('Average Unemployment Rate (%)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

# COVID-19 Period Comparison-

```python
df['Period'] = np.where(
    df['Date'] < pd.Timestamp('2020-03-01'),
    'Pre-COVID baseline',
    'COVID period'
)

period_summary = df.groupby('Period')[[
    'Estimated Unemployment Rate (%)',
    'Estimated Employed',
    'Estimated Labour Participation Rate (%)'
]].mean().round(2)

display(period_summary)

pre = period_summary.loc['Pre-COVID baseline', 'Estimated Unemployment Rate (%)']
covid = period_summary.loc['COVID period', 'Estimated Unemployment Rate (%)']
change = covid - pre
pct_change = (change / pre) * 100

print(f'Average unemployment rate change: {change:.2f} percentage points')
print(f'Relative change: {pct_change:.2f}%')
```

```python
plt.figure(figsize=(7,5))
sns.barplot(x=period_summary.index, y=period_summary['Estimated Unemployment Rate (%)'])
plt.title('Average Unemployment Rate: Pre-COVID vs COVID Period')
plt.xlabel('Period')
plt.ylabel('Average Unemployment Rate (%)')
plt.xticks(rotation=10)
plt.tight_layout()
plt.show()
```

# Highest-Unemployment States/Regions-

```python
state_avg = (
    df.groupby('Region')['Estimated Unemployment Rate (%)']
      .mean()
      .sort_values(ascending=False)
      .reset_index()
)

display(state_avg.head(10).round(2))

plt.figure(figsize=(10,6))
sns.barplot(
    data=state_avg.head(10),
    x='Estimated Unemployment Rate (%)',
    y='Region'
)
plt.title('Top 10 Regions by Average Unemployment Rate')
plt.xlabel('Average Unemployment Rate (%)')
plt.ylabel('Region')
plt.tight_layout()
plt.show()
```

# Peak Unemployment During the COVID Period-

```python
peak_state = (
    df[df['Date'] >= pd.Timestamp('2020-03-01')]
      .groupby('Region')['Estimated Unemployment Rate (%)']
      .max()
      .sort_values(ascending=False)
      .reset_index(name='Peak Unemployment Rate (%)')
)

display(peak_state.head(10).round(2))
```

# Relationship Between Unemployment and Labour Participation-

```python
plt.figure(figsize=(9,6))
sns.scatterplot(
    data=df,
    x='Estimated Labour Participation Rate (%)',
    y='Estimated Unemployment Rate (%)',
    hue='Period',
    alpha=0.7
)
plt.title('Unemployment Rate vs Labour Participation Rate')
plt.xlabel('Labour Participation Rate (%)')
plt.ylabel('Unemployment Rate (%)')
plt.tight_layout()
plt.show()

corr = df['Estimated Unemployment Rate (%)'].corr(
    df['Estimated Labour Participation Rate (%)']
)
print(f'Correlation coefficient: {corr:.3f}')
```

# Correlation Heatmap-

```python
corr_data = df[numeric_cols].corr()

plt.figure(figsize=(8,6))
sns.heatmap(corr_data, annot=True, fmt='.2f', cmap='Blues')
plt.title('Correlation Matrix')
plt.tight_layout()
plt.show()
```

# Key Insights-
The following cell generates concise findings automatically from the cleaned dataset.

```python
highest_avg_state = state_avg.iloc[0]['Region']
highest_avg_rate = state_avg.iloc[0]['Estimated Unemployment Rate (%)']
peak_row = df.loc[df['Estimated Unemployment Rate (%)'].idxmax()]

print('KEY INSIGHTS')
print('-' * 50)
print(f'1. Highest average unemployment region: {highest_avg_state} ({highest_avg_rate:.2f}%).')
print(
    f"2. Highest observed unemployment rate: "
    f"{peak_row['Estimated Unemployment Rate (%)']:.2f}% in "
    f"{peak_row['Region']} on {peak_row['Date'].strftime('%d-%b-%Y')}."
)
print(
    f'3. Average unemployment rate changed by {change:.2f} percentage points '
    f'between the pre-COVID baseline and COVID-period window.'
)
print(f'4. Relative change between the two windows: {pct_change:.2f}%.')
print(f'5. Unemployment/labour-participation correlation: {corr:.3f}.')
```
